```python
# 13.1 Write the current date as a string to the text file today.txt.
from datetime import date

text = date.today()
with open('today.txt', 'wt') as fout:
    fout.write(str(text))

# 13.2 Read the text file today.txt into the string today_string.

with open('today.txt', 'r') as fin:
    text = fin.read()

print(text)


# 13.3 Parse the date from today_string.
import time
fmt = "%Y-%m-%d"
parsed = time.strptime(text, fmt)
print(parsed)

# 15.1 Use multiprocessing to create three separate processes. Make each one wait a random number
# of seconds between zero and one, print the current time, and then exit.
import multiprocessing
import time
import random
import os
from datetime import datetime

def whoami(name):
    print("I'm %s, in process %s" % (name, os.getpid()))

def loopy(name):
    whoami(name)
    wait = random.random()
    for num in range(3):
        print("\tNumber %s of %s" % (num, 3))
        time.sleep(wait)
        now = datetime.now().strftime("%H:%M:%S.%f")
        print(now)

if __name__ == "__main__":
    whoami("main")
    p = multiprocessing.Process(target=loopy, args=("loopy",))
    p.start()
```
